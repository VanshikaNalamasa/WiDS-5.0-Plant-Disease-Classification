```python
# Week 3: Deep Learning - Plant Disease Classification
# WiDS 5.0 - CNN & Transfer Learning

# ============================================
# SECTION 1: Import Libraries
# ============================================
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
import warnings
warnings.filterwarnings('ignore')

# TensorFlow/Keras imports
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

print("="*60)
print(f"✅ TensorFlow version: {tf.__version__}")
print(f"🔧 GPU Available: {len(tf.config.list_physical_devices('GPU')) > 0}")
print("="*60)

# ============================================
# SECTION 2: Configuration
# ============================================

# Paths
DATA_PATH = '/kaggle/input/plantvillage-dataset/color'

# Image settings
IMG_HEIGHT = 128  # Using 128x128 for speed (not 224x224)
IMG_WIDTH = 128
BATCH_SIZE = 32
EPOCHS_SIMPLE = 10  # For simple CNN
EPOCHS_TRANSFER = 10  # For transfer learning

# Limit samples for faster training (remove for full training)
SAMPLES_PER_CLASS = 200  # Increase if you have time

print(f"\n📐 Image size: {IMG_HEIGHT}x{IMG_WIDTH}")
print(f"📦 Batch size: {BATCH_SIZE}")
print(f"🔄 Epochs: {EPOCHS_SIMPLE} (Simple CNN), {EPOCHS_TRANSFER} (Transfer)")
print(f"📊 Samples per class: {SAMPLES_PER_CLASS}")

# ============================================
# SECTION 3: Data Loading with ImageDataGenerator
# ============================================

print("\n" + "="*60)
print("DATA LOADING & AUGMENTATION")
print("="*60)

# Get class names and count
all_classes = sorted([d for d in os.listdir(DATA_PATH) 
                     if os.path.isdir(os.path.join(DATA_PATH, d))])
NUM_CLASSES = len(all_classes)

print(f"\n📊 Found {NUM_CLASSES} classes")

# Create temporary directory with limited samples
import shutil
import random

TEMP_DATA_PATH = '/kaggle/working/temp_data'
if os.path.exists(TEMP_DATA_PATH):
    shutil.rmtree(TEMP_DATA_PATH)
os.makedirs(TEMP_DATA_PATH)

print(f"\n⏳ Creating subset with {SAMPLES_PER_CLASS} samples per class...")

for class_name in all_classes:
    src_dir = os.path.join(DATA_PATH, class_name)
    dst_dir = os.path.join(TEMP_DATA_PATH, class_name)
    os.makedirs(dst_dir)
    
    # Get all images
    all_images = [f for f in os.listdir(src_dir) if f.endswith(('.jpg', '.JPG', '.png'))]
    
    # Sample randomly
    selected = random.sample(all_images, min(SAMPLES_PER_CLASS, len(all_images)))
    
    # Copy to temp directory
    for img in selected:
        shutil.copy(os.path.join(src_dir, img), os.path.join(dst_dir, img))

print(f"✅ Subset created!")

# Data Augmentation for training
train_datagen = ImageDataGenerator(
    rescale=1./255,              # Normalize to [0,1]
    validation_split=0.2,        # 80% train, 20% validation
    rotation_range=20,           # Rotate up to 20 degrees
    width_shift_range=0.2,       # Shift horizontally
    height_shift_range=0.2,      # Shift vertically
    horizontal_flip=True,        # Random horizontal flip
    zoom_range=0.2,              # Random zoom
    fill_mode='nearest'          # Fill missing pixels
)

# No augmentation for validation (only rescaling)
val_datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2
)

# Training generator
train_generator = train_datagen.flow_from_directory(
    TEMP_DATA_PATH,
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='training',
    shuffle=True
)

# Validation generator
val_generator = val_datagen.flow_from_directory(
    TEMP_DATA_PATH,
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='validation',
    shuffle=False
)

print(f"\n✅ Training samples: {train_generator.samples}")
print(f"✅ Validation samples: {val_generator.samples}")

# Visualize augmented samples
print("\n📸 Sample Augmented Images:")
sample_batch, sample_labels = next(train_generator)

fig, axes = plt.subplots(2, 5, figsize=(15, 6))
fig.suptitle('Data Augmentation Examples', fontsize=14, fontweight='bold')

for i, ax in enumerate(axes.flat):
    ax.imshow(sample_batch[i])
    class_idx = np.argmax(sample_labels[i])
    class_name = list(train_generator.class_indices.keys())[class_idx]
    ax.set_title(class_name.split('___')[-1][:15], fontsize=8)
    ax.axis('off')

plt.tight_layout()
plt.show()

# ============================================
# SECTION 4: Simple CNN from Scratch
# ============================================

print("\n" + "="*60)
print("STAGE 1: SIMPLE CNN (FROM SCRATCH)")
print("="*60)

def create_simple_cnn(input_shape, num_classes):
    """
    Simple CNN architecture:
    Conv -> ReLU -> MaxPool -> Conv -> ReLU -> MaxPool -> Dense -> Softmax
    """
    model = models.Sequential([
        # First Convolutional Block
        layers.Conv2D(32, (3, 3), activation='relu', input_shape=input_shape),
        layers.MaxPooling2D((2, 2)),
        layers.Dropout(0.25),
        
        # Second Convolutional Block
        layers.Conv2D(64, (3, 3), activation='relu'),
        layers.MaxPooling2D((2, 2)),
        layers.Dropout(0.25),
        
        # Third Convolutional Block
        layers.Conv2D(128, (3, 3), activation='relu'),
        layers.MaxPooling2D((2, 2)),
        layers.Dropout(0.25),
        
        # Flatten and Dense layers
        layers.Flatten(),
        layers.Dense(256, activation='relu'),
        layers.Dropout(0.5),
        layers.Dense(num_classes, activation='softmax')
    ])
    
    return model

# Create model
simple_cnn = create_simple_cnn((IMG_HEIGHT, IMG_WIDTH, 3), NUM_CLASSES)

# Compile
simple_cnn.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Model summary
print("\n📋 Simple CNN Architecture:")
simple_cnn.summary()

# Callbacks
early_stop = EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True)
reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=2, min_lr=1e-7)

# Train
print("\n🚀 Training Simple CNN...")
history_simple = simple_cnn.fit(
    train_generator,
    validation_data=val_generator,
    epochs=EPOCHS_SIMPLE,
    callbacks=[early_stop, reduce_lr],
    verbose=1
)

# Evaluate
print("\n📊 Simple CNN Evaluation:")
val_loss, val_acc = simple_cnn.evaluate(val_generator, verbose=0)
print(f"   Validation Accuracy: {val_acc*100:.2f}%")
print(f"   Validation Loss: {val_loss:.4f}")

# ============================================
# SECTION 5: Transfer Learning with MobileNetV2
# ============================================

print("\n" + "="*60)
print("STAGE 2: TRANSFER LEARNING (MobileNetV2)")
print("="*60)

def create_transfer_model(input_shape, num_classes):
    """
    Transfer Learning using MobileNetV2:
    - Load pre-trained weights from ImageNet
    - Freeze base layers
    - Add custom classification head
    """
    # Load MobileNetV2 without top layer
    base_model = MobileNetV2(
        input_shape=input_shape,
        include_top=False,
        weights='imagenet'
    )
    
    # Freeze base model
    base_model.trainable = False
    
    # Add custom head
    model = models.Sequential([
        base_model,
        layers.GlobalAveragePooling2D(),
        layers.Dropout(0.5),
        layers.Dense(256, activation='relu'),
        layers.Dropout(0.3),
        layers.Dense(num_classes, activation='softmax')
    ])
    
    return model, base_model

# Create model
transfer_model, base_model = create_transfer_model((IMG_HEIGHT, IMG_WIDTH, 3), NUM_CLASSES)

# Compile
transfer_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print("\n📋 Transfer Learning Model:")
print(f"   Base model: MobileNetV2 (frozen)")
print(f"   Trainable params: {transfer_model.count_params() - base_model.count_params()}")
print(f"   Total params: {transfer_model.count_params()}")

# Train
print("\n🚀 Training Transfer Learning Model...")
history_transfer = transfer_model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=EPOCHS_TRANSFER,
    callbacks=[early_stop, reduce_lr],
    verbose=1
)

# Evaluate
print("\n📊 Transfer Learning Evaluation:")
val_loss_transfer, val_acc_transfer = transfer_model.evaluate(val_generator, verbose=0)
print(f"   Validation Accuracy: {val_acc_transfer*100:.2f}%")
print(f"   Validation Loss: {val_loss_transfer:.4f}")

# ============================================
# OPTIONAL: Fine-tuning (Unfreeze some layers)
# ============================================

print("\n" + "="*60)
print("STAGE 3: FINE-TUNING (OPTIONAL)")
print("="*60)

print("\n🔧 Unfreezing last 20 layers of base model for fine-tuning...")

# Unfreeze the last layers
base_model.trainable = True
for layer in base_model.layers[:-20]:
    layer.trainable = False

# Recompile with lower learning rate
transfer_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.0001),  # Much lower LR
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print(f"   Trainable params after unfreezing: {sum([tf.size(w).numpy() for w in transfer_model.trainable_weights])}")

# Fine-tune
print("\n🚀 Fine-tuning...")
history_finetune = transfer_model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=5,  # Just a few epochs
    callbacks=[early_stop],
    verbose=1
)

# Final evaluation
val_loss_final, val_acc_final = transfer_model.evaluate(val_generator, verbose=0)
print(f"\n📊 After Fine-tuning:")
print(f"   Validation Accuracy: {val_acc_final*100:.2f}%")
print(f"   Validation Loss: {val_loss_final:.4f}")

# ============================================
# SECTION 6: Training History Visualization
# ============================================

print("\n" + "="*60)
print("TRAINING CURVES")
print("="*60)

fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Simple CNN - Accuracy
axes[0, 0].plot(history_simple.history['accuracy'], label='Train', linewidth=2)
axes[0, 0].plot(history_simple.history['val_accuracy'], label='Validation', linewidth=2)
axes[0, 0].set_title('Simple CNN - Accuracy', fontweight='bold', fontsize=12)
axes[0, 0].set_xlabel('Epoch')
axes[0, 0].set_ylabel('Accuracy')
axes[0, 0].legend()
axes[0, 0].grid(alpha=0.3)

# Simple CNN - Loss
axes[0, 1].plot(history_simple.history['loss'], label='Train', linewidth=2)
axes[0, 1].plot(history_simple.history['val_loss'], label='Validation', linewidth=2)
axes[0, 1].set_title('Simple CNN - Loss', fontweight='bold', fontsize=12)
axes[0, 1].set_xlabel('Epoch')
axes[0, 1].set_ylabel('Loss')
axes[0, 1].legend()
axes[0, 1].grid(alpha=0.3)

# Transfer Learning - Accuracy
axes[1, 0].plot(history_transfer.history['accuracy'], label='Train', linewidth=2)
axes[1, 0].plot(history_transfer.history['val_accuracy'], label='Validation', linewidth=2)
axes[1, 0].set_title('Transfer Learning - Accuracy', fontweight='bold', fontsize=12)
axes[1, 0].set_xlabel('Epoch')
axes[1, 0].set_ylabel('Accuracy')
axes[1, 0].legend()
axes[1, 0].grid(alpha=0.3)

# Transfer Learning - Loss
axes[1, 1].plot(history_transfer.history['loss'], label='Train', linewidth=2)
axes[1, 1].plot(history_transfer.history['val_loss'], label='Validation', linewidth=2)
axes[1, 1].set_title('Transfer Learning - Loss', fontweight='bold', fontsize=12)
axes[1, 1].set_xlabel('Epoch')
axes[1, 1].set_ylabel('Loss')
axes[1, 1].legend()
axes[1, 1].grid(alpha=0.3)

plt.tight_layout()
plt.show()

# ============================================
# SECTION 7: Confusion Matrix
# ============================================

print("\n" + "="*60)
print("CONFUSION MATRIX")
print("="*60)

# Get predictions
print("\n⏳ Generating predictions...")
val_generator.reset()
y_pred_probs = transfer_model.predict(val_generator, verbose=0)
y_pred = np.argmax(y_pred_probs, axis=1)
y_true = val_generator.classes

# Confusion matrix
cm = confusion_matrix(y_true, y_pred)

# Plot (show subset if too many classes)
plt.figure(figsize=(14, 12))

if NUM_CLASSES > 20:
    # Show top 15 classes
    top_classes_idx = np.argsort([np.sum(y_true == i) for i in range(NUM_CLASSES)])[-15:]
    cm_subset = cm[np.ix_(top_classes_idx, top_classes_idx)]
    class_names_subset = [all_classes[i].split('___')[-1][:20] for i in top_classes_idx]
    
    sns.heatmap(cm_subset, annot=True, fmt='d', cmap='Blues',
                xticklabels=class_names_subset,
                yticklabels=class_names_subset,
                cbar_kws={'label': 'Count'})
    plt.title('Confusion Matrix (Top 15 Classes)', fontsize=14, fontweight='bold')
else:
    class_names_display = [c.split('___')[-1][:20] for c in all_classes]
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=class_names_display,
                yticklabels=class_names_display,
                cbar_kws={'label': 'Count'})
    plt.title('Confusion Matrix (All Classes)', fontsize=14, fontweight='bold')

plt.xlabel('Predicted Label', fontweight='bold')
plt.ylabel('True Label', fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

# ============================================
# SECTION 8: Performance Comparison
# ============================================

print("\n" + "="*60)
print("🏆 FINAL COMPARISON")
print("="*60)

results = {
    'Simple CNN': val_acc * 100,
    'Transfer Learning': val_acc_transfer * 100,
    'Transfer + Fine-tune': val_acc_final * 100
}

print("\n📊 Model Performance:")
for model_name, acc in results.items():
    print(f"   {model_name:25s}: {acc:>6.2f}%")

# Visualization
plt.figure(figsize=(12, 6))
models_list = list(results.keys())
accs_list = list(results.values())

colors = ['#3498db', '#2ecc71', '#f39c12']
bars = plt.bar(models_list, accs_list, color=colors, edgecolor='black', linewidth=2)

# Add value labels
for bar, acc in zip(bars, accs_list):
    plt.text(bar.get_x() + bar.get_width()/2, acc + 1,
             f'{acc:.1f}%', ha='center', va='bottom',
             fontweight='bold', fontsize=12)

plt.ylabel('Validation Accuracy (%)', fontweight='bold', fontsize=12)
plt.title('Week 3: Deep Learning Model Comparison', fontweight='bold', fontsize=14)
plt.ylim(0, max(accs_list) + 10)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# ============================================
# SECTION 9: Comparison with Week 2 Baseline
# ============================================

print("\n" + "="*60)
print("📈 WEEK 2 vs WEEK 3 COMPARISON")
print("="*60)

# Assume Week 2 baseline was ~50% (adjust based on your actual result)
week2_baseline = 65.8 # UPDATE THIS with your actual Week 2 best result

print(f"\n📊 Progress:")
print(f"   Week 2 Baseline (Random Forest): {week2_baseline:.1f}%")
print(f"   Week 3 Simple CNN:               {val_acc*100:.1f}% (+{val_acc*100 - week2_baseline:.1f}%)")
print(f"   Week 3 Transfer Learning:        {val_acc_transfer*100:.1f}% (+{val_acc_transfer*100 - week2_baseline:.1f}%)")
print(f"   Week 3 Fine-tuned:               {val_acc_final*100:.1f}% (+{val_acc_final*100 - week2_baseline:.1f}%)")

improvement = val_acc_final*100 - week2_baseline

if improvement > 20:
    print(f"\n✅ EXCELLENT! {improvement:.1f}% improvement justifies deep learning!")
elif improvement > 10:
    print(f"\n✅ GOOD! {improvement:.1f}% improvement - deep learning adds value")
else:
    print(f"\n⚠️ Modest {improvement:.1f}% improvement - may need more tuning")

# ============================================
# SECTION 10: Key Insights & Reflection
# ============================================

print("\n" + "="*60)
print("🔍 KEY INSIGHTS & LEARNINGS")
print("="*60)

print("\n1️⃣ Why Did Transfer Learning Win?")
print("   • Pre-trained on ImageNet (1.4M images)")
print("   • Already learned: edges, textures, shapes")
print("   • We just taught it: 'this texture = late blight'")

print("\n2️⃣ What Did CNNs Learn That Classical ML Couldn't?")
print("   • Spatial patterns (pixel neighbors matter)")
print("   • Hierarchical features (edges → textures → spots)")
print("   • Translation invariance (lesion anywhere on leaf)")

print("\n3️⃣ Where Does the Model Still Fail?")
print("   • Similar diseases (e.g., different types of blight)")
print("   • Poor lighting or image quality")
print("   • Minority classes with fewer training samples")

print("\n4️⃣ How Could We Improve Further?")
print("   • More aggressive data augmentation")
print("   • Larger image size (224x224 instead of 128x128)")
print("   • Ensemble multiple models (vote)")
print("   • Attention mechanisms to focus on disease spots")

print("\n" + "="*60)
print("✅ WEEK 3 COMPLETE!")
print("="*60)
print("\n📚 Summary:")
print(f"   • Built CNN from scratch: {val_acc*100:.1f}%")
print(f"   • Applied Transfer Learning: {val_acc_transfer*100:.1f}%")
print(f"   • Fine-tuned for: {val_acc_final*100:.1f}%")
print(f"   • Beat Week 2 baseline by: +{improvement:.1f}%")
print("\n🎉 You've completed the full ML pipeline!")
print("="*60)
```
